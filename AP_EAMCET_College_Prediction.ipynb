```python
missing_by_branch = df.groupby("branch_code")["OC_EWS_BOYS"].agg(
    total="size",
    missing=lambda x: x.isna().sum()
)

missing_by_branch["missing_pct"] = (
    missing_by_branch["missing"] / missing_by_branch["total"] * 100
)

print(missing_by_branch.sort_values("missing_pct", ascending=False))
```

```python
id_cols = [
    "INSTCODE",
    "NAME OF THE INSTITUTION",
    "TYPE",
    "INST_REG",
    "DIST",
    "PLACE",
    "COED",
    "AFFL.",
    "ESTD",
    "A_REG",
    "branch_code",
    "COLLFEE"
]

df_long = df.melt(
    id_vars=id_cols,
    value_vars=cutoff_cols,
    var_name="Category_Gender",
    value_name="Cutoff"
)

print(df_long.shape)
print(df_long.head())
```

```python
df_long[["Category", "Gender"]] = df_long["Category_Gender"].str.rsplit(
    "_", n=1, expand=True
)

print(df_long[["Category_Gender", "Category", "Gender"]].drop_duplicates())
```

```python
df_long = df_long.dropna(subset=["Cutoff"]).reset_index(drop=True)

print("Shape:", df_long.shape)
print("Missing Cutoff:", df_long["Cutoff"].isna().sum())
```
