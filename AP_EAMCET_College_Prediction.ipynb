```python
cutoff_cols = df.columns[12:30]

for col in cutoff_cols:
    values = df[col].dropna().astype(str)
    non_numeric = values[~values.str.fullmatch(r"\d+")]

    if len(non_numeric) > 0:
        print(col, "->", non_numeric.unique())
```

```python
for col in cutoff_cols:
    blank_count = (df[col] == ' ').sum()
    print(f"{col}: {blank_count}")
```

```python
print(df.iloc[0].to_string())
```

```python
print("Unique branch codes:", df["branch_code"].nunique())

print("\nBranch codes:")
print(df["branch_code"].unique())
```

```python
branch_counts = df["branch_code"].value_counts()

print(branch_counts)
print("Minimum records for a branch:", branch_counts.min())
print("Maximum records for a branch:", branch_counts.max())
```

```python
college_branch = df.groupby(
    ["INSTCODE", "branch_code"]
).size()

print("Total college-branch combinations:", len(college_branch))
print("Total dataset rows:", len(df))
```

```python
duplicates = college_branch[college_branch > 1]

print("College-branch combinations appearing more than once:",
      len(duplicates))

print(duplicates.head(20))
```

```python
df[
    (df["INSTCODE"] == "ADTPPU") &
    (df["branch_code"] == "AGR")
].T
```

```python
college_branch_region = df.groupby(
    ["INSTCODE", "branch_code", "A_REG"]
).size()

print("Unique college-branch-region combinations:",
      len(college_branch_region))

print("Repeated combinations:",
      (college_branch_region > 1).sum())
```

```python
areg_counts = df["A_REG"].value_counts(dropna=False)

print(areg_counts)
```

```python
cutoff_cols = df.columns[12:30]

categories = sorted({
    col.rsplit("_", 1)[0]
    for col in cutoff_cols
})

print("Categories:", categories)
print("Number of categories:", len(categories))
```

```python
genders = sorted({
    col.rsplit("_", 1)[1]
    for col in cutoff_cols
})

print("Genders:", genders)
print("Number of genders:", len(genders))
```

```python
print("COLLFEE dtype:", df["COLLFEE"].dtype)

print("\nMissing values:")
print(df["COLLFEE"].isna().sum())

print("\nUnique sample values:")
print(df["COLLFEE"].unique()[:20])
```

```python
non_numeric_fee = df[
    ~df["COLLFEE"].apply(lambda x: str(x).strip().isdigit())
]

print("Non-numeric COLLFEE values:")
print(non_numeric_fee["COLLFEE"].unique())
print("Count:", len(non_numeric_fee))
```

```python
categorical_cols = ["TYPE", "INST_REG", "COED", "AFFL.", "A_REG"]

for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].value_counts())
```
